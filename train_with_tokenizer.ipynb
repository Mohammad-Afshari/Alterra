## Create Batching Pipeline & Loss Estimator And Define Optimizer & Model

```python
def get_batch(data):
    """takes a random batch of input x and target y from data"""

    maxStartOfBatch = len(data) - window_size
    startOfBatchSamples = torch.randint(maxStartOfBatch, (batch_size,))
    x = torch.stack([ data[i:i + window_size] for i in startOfBatchSamples] )
    y = torch.stack([ data[i + 1:i + window_size + 1] for i in startOfBatchSamples] )
    return x ,y
```

```python
model = Model(n_embed, window_size, vocab_size, num_heads, num_transformer_blocks, dropout)

optimizer = torch.optim.AdamW(model.parameters(), lr=learning_rate)
print(sum(p.numel() for p in model.parameters()) / 1e6, 'M parameters')
```

```python
@torch.no_grad()
def estimate_loss(train_data, val_data):
    # this function estimates both train and eval loss
    out = {}

    model.eval()

    for split in ['train', 'val']:
        losses = torch.zeros(eval_iters)
        for k in range(eval_iters):

            if split == 'train':
                X, Y = get_batch(train_data)
            else:
                X, Y = get_batch(val_data)

            logits, loss = model(X, Y)
            losses[k] = loss.item()

        out[split] = losses.mean()

    model.train()

    return out
```

## Training Loop

```python
n = int(0.9*len(data))
pretrain_train_data = data[:n]
pretrain_val_data = data[n:]

train_losses = []
val_losses = []
loss_steps = []


fig, ax = plt.subplots(1,2)
ax[0].set_title("Train Losses")
ax[1].set_title("Eval Losses")

n_steps = int(pretrain_iters/eval_interval)
for step in range(n_steps):
    print(f'Step {step*eval_interval}/{pretrain_iters}', end=" ")
    pbar = tqdm(range(eval_interval))
    for iter in pbar:
        # For Monitoring Porpuse
        if iter == eval_interval-1:
            losses = estimate_loss(pretrain_train_data, pretrain_val_data)
            pbar.set_postfix(loss=f"Train: {losses['train']:.4f}, Eval: {losses['val']:.4f}")
            # print(f'step {iter}:\n    train_loss {losses['train']:.4f}\n    eval_loss {losses['eval']:.4f} \n\n')

            train_losses.append(float(losses['train']))
            val_losses.append(float(losses['val']))
            loss_steps.append( iter + (step*eval_interval) )

        # Training Step
        xb, yb = get_batch(pretrain_train_data)

        logits, loss = model(xb, yb)
        optimizer.zero_grad(set_to_none=True)
        loss.backward()
        optimizer.step()

print(f"""


Training Phase Has Concluded.

Final Loss:
    Train: {train_losses[-1]}
    Val: {val_losses[-1]}

Model Hyperparameters:
    n embed: {n_embed}
    num heads: {num_heads}
    num blocks: {num_transformer_blocks}
    window size: {window_size}
    dropout value: {dropout}
    learning rate: {learning_rate}
    num parameters: {sum(p.numel() for p in model.parameters()) / 1e6, 'M parameters'}

Step_Loss Curve:

""")

ax[0].plot(loss_steps, train_losses)
ax[1].plot(loss_steps, val_losses)
plt.show()
```
